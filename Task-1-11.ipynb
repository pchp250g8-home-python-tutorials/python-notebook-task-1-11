При каких целых значениях n выражение (3 * n - 1) / (n + 2) является натуральным чиcлом?

```python
from numpy import array
from sympy import pdiv
from sympy import apart
from sympy import divisors
from sympy import solve
from sympy.abc import n
from sympy import Eq


ans = []; c = 0
expr1 = pdiv(3 * n - 1, n + 2)
expr2 = apart((3 * n - 1) / (n + 2))
print(expr1,expr2)
divs = array(divisors(7))
print(divs, -divs)
divs=divs.tolist() + (-divs).tolist()
for i in divs:
    eqn = Eq(-7 / (n + 2), i)
    res = solve(eqn, n)
    print("Equation N ",(c + 1),eqn.lhs,"=",eqn.rhs,"Solution:",res)
    m = expr2.subs({n:res[0]})
    if(m > 0):
        ans.append(res[0])
    c+=1
ans.sort()
print(ans)
```
