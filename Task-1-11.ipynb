При каких целых значениях n выражение (3 * n - 1) / (n + 2) является натуральным чиcлом?

```python
# --способ № 1---
from numpy import array
from sympy import pdiv
from sympy import apart
from sympy import divisors
from sympy import solve
from sympy.abc import n
from sympy import Eq
from sympy import numer


ans = []; c = 0
expr1 = (3 * n - 1) / (n + 2)
expr2 = apart(expr1)
n_len = len(expr2.args)
expr3 = expr2.args[n_len-1]
expr4 = numer(expr3)
print("Source fract:",expr1)
print("Converted fract:",expr2)
pos_divs = divisors(expr4)
neg_divs = [-x for x in pos_divs]
divs = pos_divs + neg_divs
print("Divisors of the numerator:", divs)
for i in divs:
    eqn = Eq(expr3, i)
    res = solve(eqn, n)
    s_eqn = str(eqn.lhs) + " = " + str(eqn.rhs)
    print("Equation N ",(c + 1),s_eqn,"Solution:",res)
    m = expr2.subs({n:res[0]})
    if(m > 0):
        ans.append(res[0])
    c+=1
ans.sort()
print("Answer:",ans)
```
